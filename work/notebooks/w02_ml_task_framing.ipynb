## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

ML Task Type: Ranking

My lane is **Search Ranking and Search Relevance Prediction**.

I frame this as a **ranking** task because the practical goal is to order content pages by their priority for search-performance review. Instead of making only a yes/no prediction, the ML system can assign a priority score to each page and produce a ranked list.

The ranking output can support a content team's decision about which pages should be reviewed or considered for refresh first. This keeps the ML output connected to a real action rather than treating model prediction as the final goal.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

I would predict whether a content page shows an observed decline in search performance.

The target is a proxy for **content-performance deterioration**, not a direct measurement of Google's relevance judgment. The label comes from an observed outcome in the dataset rather than from a subjective manual definition.

For the ranking task, the model would estimate which pages are more likely to belong to the declining group and use that estimate to prioritize pages for review.

I will check the exact target definition and available features before treating it as the final modeling target, especially to avoid using information that would only be available after the outcome occurs.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

My primary success metric will be **Precision@50**.

Precision@50 measures the proportion of the 50 highest-ranked pages that are actually in the target group.

This metric matches the practical decision because a content team has limited time and may only be able to review a small number of pages. A useful ranking system should therefore place a high proportion of genuinely high-priority pages near the top.

I will compare the ML ranking against a simple fixed-rule baseline rather than assuming that ML is automatically better.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule could prioritize pages using one manually selected condition, such as a CTR threshold or a particular average-position threshold. However, search performance can depend on several characteristics of a page and its search environment at the same time.

An ML model can learn combinations of multiple observed signals rather than relying on one manually chosen threshold. This makes it possible to produce a more flexible ranking of pages for review.

However, I will not assume that ML is automatically better. I will compare the ML approach with a simple fixed-rule baseline using the same evaluation metric. If the ML approach does not provide meaningful improvement, the simpler rule may be preferable.
